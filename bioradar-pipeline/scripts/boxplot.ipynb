# Create a boxplot from a seven number summary

```python
import pandas as pd
```

```python
df = pd.read_table("../plots/paired-end-demux.qzv/forward-seven-number-summaries.tsv")
df
```

```python
%matplotlib inline
def create_box(counter,col):
    if counter %5 != 0:
        counter = ''
        
    box = {
        'label' : counter,
        'whislo': col[0],    # Bottom whisker position
        'q1'    : col[1],    # First quartile (25th percentile)
        'med'   : col[2],    # Median         (50th percentile)
        'q3'    : col[3],    # Third quartile (75th percentile)
        'whishi': col[4],    # Top whisker position
        'fliers': []        # Outliers
        }
    return box
```

```python
#box = df.iloc[2:7,1:]
#box = df.iloc[3:6,1:]
#min, 25, 50,75, max
box_data = df.iloc[[1,3,4,5,7],1:]
```

```python
boxes = []
for ix, column in enumerate(box_data):
    col = box_data[column].values
    box = create_box(ix, col)
    boxes.append(box)
```

```python
boxes
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()
ax.bxp(boxes, showfliers=False)
ax.set_title("Quality plot forward read")
ax.set_ylabel("Quality Scores")
ax.set_xlabel("Read base position")
plt.savefig("boxplot.png")
plt.close()
```
