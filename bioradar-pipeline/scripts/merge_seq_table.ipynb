```python
from Bio import SeqIO
# merge the asv table and sequences
dna_dict = {}
with open('/media/dwheeler/spinner/Linux_space/projects/bioinformatics/202302mduncan/qiime2/12S/asvs/dna-sequences.fasta') as rf:
    for rec in SeqIO.parse(rf, 'fasta'):
        assert rec.id not in dna_dict
        dna_dict[rec.id] = str(rec.seq)
```

```python
import csv
with open('../results/asv_count_table_sequences.csv', 'w') as wf:
    csv_writer = csv.writer(wf)
    with open('/media/dwheeler/spinner/Linux_space/projects/bioinformatics/202302mduncan/qiime2/12S/asvs/asv_count_tax.tsv') as rf:
        header = next(rf).strip().split('\t') + ['asv_seq']
        csv_writer.writerow(header)
        for row in rf:
            row = row.split('\t')
            row.append(dna_dict[row[0]])
            csv_writer.writerow(row)
```
