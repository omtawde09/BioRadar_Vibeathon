## Create a table of raw species counts across samples

```python
import pandas as pd

count_table = pd.read_table('../qiime2/12S/asvs/asv_count_tax.tsv')
count_table.head()
```

```python
#Drop the hash
count_table.drop('Feature.ID', inplace=True, axis=1)
count_table.head()
```

```python
count_table['midori_Confidence'].median()
```

```python
# drop confidence
count_table.drop('midori_Confidence', axis=1, inplace=True)
count_table.head()
```

```python
def fmt_taxonomy(taxonomy):
    tree = taxonomy.split(';')
    return tree[-1]

count_table['species'] = count_table['midori_Taxon'].apply(fmt_taxonomy)
```

```python
count_table.drop('midori_Taxon', axis=1, inplace=True)
count_table.head()
```

```python
by_species = count_table.groupby('species').sum()
by_species.loc['sum'] = by_species.sum(axis=0)
by_species.to_csv('../results/raw_species_counts.csv')
by_species.tail()
```

## Create a table of normalised species counts across samples
So I checked, and all OTUs could be assigned a taxa (at some level), so column sums actually repressent read count sums (ie there are no unallocated OUTs whose counts are dropped from the table). So all i have to do is divide each value by the column sum to give a relative abundance.

```python
# drop the sum row as that will effect the column counts
by_species = by_species.iloc[:-1,:]
# divide by col sum
by_species_rel = by_species.div(by_species.sum(axis=0), axis=1)
by_species_rel.head()
```

```python
by_species_rel.loc['sum'] = by_species_rel.sum(axis=0)
by_species_rel
by_species_rel.to_csv('../results/relative_species_counts.csv')
```
